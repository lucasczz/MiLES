```python
import pandas as pd
from src.data.metrics import get_metrics
import numpy as np
```

```python
# Get best hyperparameters for TUL methods
df = pd.read_json("../reports/method_hparams_TUL.jsonl", lines=True)
metrics = ["top_1_accuracy"]
get_metrics(df, metrics)
hparams = [
    "loc_levels",
    "discretization_rows",
    "grow_factor",
    "embedding_type",
    "replay_buffer_type",
    "embedding_weight_factor",
    "replay_batch_size",
    "lr",
    "optimizer_cls"
]
idcs_max = df.groupby(["embedding_type", "replay_buffer_type", "optimizer_cls"])[
    "top_1_accuracy"
].idxmax()
df.loc[idcs_max][hparams + metrics]
```

```python
df = pd.read_json("../reports/method_hparams_DP.jsonl", lines=True)
df["mean_dist"] = df["pred_dists"].apply(lambda x: np.mean(x))

hparams = [
    "loc_levels",
    "discretization_rows",
    "grow_factor",
    "embedding_type",
    "replay_buffer_type",
    "embedding_weight_factor",
    "replay_batch_size",
    'lr'
]

min_idcs = df.groupby(["embedding_type", "replay_buffer_type"])["mean_dist"].idxmin()
df.loc[min_idcs][hparams + ["mean_dist"]]
```
